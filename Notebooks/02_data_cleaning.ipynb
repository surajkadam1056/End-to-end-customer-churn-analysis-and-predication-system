Data Cleaning:

In this phase, we clean the dataset by handling missing values, removing duplicates and preparing data for analysis and modeling.


```python
import pandas as pd 
import numpy as np
df=pd.read_csv("ecommerce_customer_churn_dataset.csv")
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
#check missing values:
df.isnull().sum().sort_values(ascending=False)
```

```python
#Missing values handle
```

```python
num_cols = df.select_dtypes(include=["int64","float64"]).columns
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
cat_cols = df.select_dtypes(include=["object"]).columns

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
#dublicate remove 
df.duplicated().sum() 
df.drop_duplicates(inplace=True)
```

```python
df.dtypes
```

```python
#data type fix          # why choose only: because float hai but actually count hona chahiye
df["Age"] = df["Age"].astype(int)
df["Total_Purchases"] = df["Total_Purchases"].astype(int)
```

```python
#column renaming 
df.columns = df.columns.str.lower().str.replace(" ","_")
```

```python
#Encode categorical Data
df = pd.get_dummies(df, drop_first=True)
df.isnull().sum()
```

```python
df.info()
```

```python
#Save cleaned data
df.to_csv("cleaned_ecommerce_customer_churn_dataset.csv", index=False)
```

```python
Insights:

- Missing values in numerical columns were filled using mean.
- Missing values in categorical columns were handled using 'Unknown'.
- Duplicate records were checked and removed.
- Column names were standardized for consistency.
```
